# Handling Missing Data

There are several **types of methods to handle missing data**, each suited for different situations depending on the amount of missingness, the type of data, and the reason behind the missing values.

Here's a categorized list of the **main ways to handle missing data**:

---

## 🔹 **1. Deletion Methods**

### a. **Listwise Deletion (Complete Case Analysis)**
- Removes entire rows with any missing value.
- **Best for**: When missingness is **MCAR** and the percentage of missing data is small.
- **Risk**: Can lead to loss of information or bias.

### b. **Pairwise Deletion**
- Uses all available data for each analysis (e.g., correlation between two variables even if others are missing).
- **Best for**: Preserving data where possible.
- **Risk**: Complicates analysis and can be inconsistent.

---

## 🔹 **2. Imputation Methods**

### a. **Mean / Median / Mode Imputation**
- Replace missing values with the mean/median/mode of the column.
- **Best for**: Numerical data, when data is **MCAR** or **MAR**.
- **Risk**: Can reduce variance and distort distributions.

### b. **Constant Value Imputation**
- Fill with a placeholder (e.g., 0, -999, "Unknown").
- **Best for**: Categorical data or flagging missingness.
- **Risk**: May mislead model if not handled carefully.

### c. **Forward/Backward Fill (Time Series)**
- Fill missing values with the previous or next known value.
- **Best for**: Time series data.
- **Risk**: Assumes stability over time.

---

## 🔹 **3. Advanced Imputation Techniques**

### a. **KNN (K-Nearest Neighbors) Imputation**
- Fills missing values using the mean (or mode) of **K similar rows**.
- **Best for**: MAR data with patterns.
- **Risk**: Computationally expensive, sensitive to scaling.

### b. **Regression Imputation**
- Predicts missing values using a regression model trained on other features.
- **Best for**: MAR data with strong correlations.
- **Risk**: May underestimate standard errors.

### c. **MICE (Multiple Imputation by Chained Equations)**
- Fills missing values multiple times using models, then pools results.
- **Best for**: MAR, complex datasets.
- **Risk**: More complex to implement.

### d. **Iterative Imputer (e.g., from sklearn)**
- Similar to MICE, models each feature with missing values as a function of other features.
- **Smart choice for**: Continuous data with inter-feature dependencies.

---

## 🔹 **4. Model-Based Approaches**

- **Tree-based methods**, like Random Forest, can handle missing values internally.
- **XGBoost**, for example, splits data based on missingness.
- **Best for**: Machine learning tasks where imputing might not be necessary.

---

## 🔹 **5. Indicator for Missingness**

- Create a new feature: e.g., `Age_missing = 1 if Age is missing, else 0`
- Combine with imputation.
- **Best for**: When missingness itself is informative (MNAR).

---

### ✅ Choosing the Right Method

| Condition                  | Recommended Strategy                        |
|---------------------------|---------------------------------------------|
| MCAR, small amount        | Listwise Deletion or Mean Imputation        |
| MAR                       | KNN, Regression, or MICE                    |
| MNAR                      | Model missingness directly, use indicators  |
| Categorical Data          | Mode or Constant ("Unknown") Imputation     |
| Time Series               | Forward/Backward Fill                       |
