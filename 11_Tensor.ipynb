### What is a Tensor?

A **tensor** is a **generalized multi-dimensional array**. It is the data structure used to represent data in machine learning models.

At its core:
- A **scalar** is a 0D tensor.
- A **vector** is a 1D tensor.
- A **matrix** is a 2D tensor.
- Higher-dimensional arrays (like a stack of matrices) are 3D, 4D, or even nD tensors.

So, **tensors = containers for data**.

---

### 🔍 Tensor Examples

| Tensor Type | Example                | Shape        | Description                      |
|-------------|------------------------|--------------|----------------------------------|
| Scalar      | `5`                    | `()`         | A single number                  |
| Vector      | `[5, 10, 15]`          | `(3,)`       | 1D array of 3 numbers            |
| Matrix      | `[[1, 2], [3, 4]]`     | `(2, 2)`     | 2D grid of numbers               |
| 3D Tensor   | `[[[1], [2]], [[3], [4]]]` | `(2, 2, 1)` | Tensor with 3 dimensions         |

---

### Tensor Dimensions (a.k.a. Rank)

- **Rank**: Number of dimensions (or axes)
- A 2D tensor has rank 2, a 3D tensor has rank 3, and so on.

---

### ⚙️ Why Tensors in Machine Learning?

Tensors are super important because they:
1. **Represent input data** (images, text, audio).
2. **Flow through models** (hence *TensorFlow*).
3. Enable **parallel computation** on GPUs.

---

### 🚀 Tensors in Deep Learning

In deep learning:
- An image is usually represented as a **3D tensor**: `(height, width, channels)`
- A batch of images becomes a **4D tensor**: `(batch_size, height, width, channels)`

In PyTorch or TensorFlow, when you pass data through a neural network, you’re passing **tensors**.

---

### 🧠 Code Example (using PyTorch)

```python
import torch

# Create a 2D tensor (matrix)
tensor_2d = torch.tensor([[1, 2], [3, 4]])
print(tensor_2d)
print("Shape:", tensor_2d.shape)
print("Rank (dim):", tensor_2d.dim())
```

---

### 🧵 Summary

- A **tensor is a container for numbers**, used in ML and DL.
- They generalize scalars, vectors, and matrices.
- Tensors are **the core data structure** in libraries like TensorFlow and PyTorch.
